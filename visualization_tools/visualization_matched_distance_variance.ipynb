```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1) Load your results CSV
df = pd.read_csv(r"Q:\Model Projects\Freeway Network Managed Lanes\matched_distance_variance/MTC_connector_variances.csv")

# 2) Pull out the variance column and drop NaNs
data = df["connector_length_variance"].dropna()
data = data[data > 0]

# 3) Compute the 90th, 95th, and 99th percentiles
q90, q95, q99 = data.quantile([0.90, 0.95, 0.99])
print(f"90th percentile = {q90:.3f}")
print(f"95th percentile = {q95:.3f}")
print(f"99th percentile = {q99:.3f}")

# 4) Build weights for proportions
weights = np.ones_like(data) / len(data)

# 5) Plot
fig, ax = plt.subplots(figsize=(10, 4))

# a) Histogram of proportions, zoomed to 0–99th pctile
ax.hist(
    data,
    bins=100,
    weights=weights,
    range=(0, q99),
    edgecolor="white",
    alpha=0.75
)

# b) Percentile lines with labels for the legend
ax.axvline(q90, color="C1", linestyle="--", linewidth=2, label=f"90th pct ({q90:.0f})")
ax.axvline(q95, color="C2", linestyle="--", linewidth=2, label=f"95th pct ({q95:.0f})")
ax.axvline(q99, color="C3", linestyle="--", linewidth=2, label=f"99th pct ({q99:.0f})")

# c) Labels, title, legend
ax.set_xlabel("Connector-length Variance (m²)")
ax.set_ylabel("Proportion of Trips")
ax.set_title(
    "MTC_Histogram of Connector-length Variances\n"
    "(bars sum to 1 ⇒ proportions; zoomed to 0–99th percentile)"
)
ax.legend(title="Percentiles")

plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# load & prep
df   = pd.read_csv(r"Q:\Model Projects\Freeway Network Managed Lanes\matched_distance_variance/SFCTA_connector_variances.csv")
data = df["connector_length_variance"].dropna()
data = data[data > 0]
q90, q95, q99 = data.quantile([0.90, 0.95, 0.99])

# restrict to 0–95th
mask = (data >= 0) & (data <= q95)
zoomed = data[mask]

# now weights sum to 1 *over the zoomed subset*
weights_in = np.ones_like(zoomed) / len(zoomed)

fig, ax = plt.subplots(figsize=(10,4))

ax.hist(
    zoomed,
    bins=50,
    weights=weights_in,
    range=(0, q95),
    edgecolor="white",
    alpha=0.75
)
ax.axvline(q90, color="C1", ls="--", lw=2, label=f"90th pct ({q90:.0f})")
ax.axvline(q95, color="C2", ls="--", lw=2, label=f"95th pct ({q95:.0f})")

ax.set_xlabel("Connector-length Variance (m²)")
ax.set_ylabel("Proportion of trips (within 0–95th pctile)")
ax.set_title("Zoomed Histogram (0–95th pctile)")
ax.legend(title="Percentiles")

plt.tight_layout()
plt.show()
```

```python
## Log version

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 1) Load your results CSV
df = pd.read_csv(r"Q:\Model Projects\Freeway Network Managed Lanes\matched_distance_variance/MTC_connector_variances.csv")

# 2) Extract and clean the variance column
data = df["connector_length_variance"].dropna()
data = data[data > 0]    # drop zeros so log() is defined

# 3) Compute raw percentiles (for labeling)...
q90_raw, q95_raw, q99_raw = data.quantile([0.90, 0.95, 0.99])
print(f"90th raw pct = {q90_raw:.3f}")
print(f"95th raw pct = {q95_raw:.3f}")
print(f"99th raw pct = {q99_raw:.3f}")

# ...and their log‐values (for plot positions)
log_data   = np.log(data)
q90_log, q95_log, q99_log = log_data.quantile([0.90, 0.95, 0.99])

# 4) Build weights so bars sum to 1
weights = np.ones_like(log_data) / len(log_data)

# 5) Plot
fig, ax = plt.subplots(figsize=(10, 4))

# a) Histogram of log‐values (proportions)
ax.hist(
    log_data,
    bins=100,
    weights=weights,
    edgecolor="white",
    alpha=0.75
)

# b) Percentile lines at the log‐thresholds
ax.axvline(q90_log, color="C1", linestyle="--", linewidth=2,
           label=f"90th pct (raw {q90_raw:.0f})")
ax.axvline(q95_log, color="C2", linestyle="--", linewidth=2,
           label=f"95th pct (raw {q95_raw:.0f})")
ax.axvline(q99_log, color="C3", linestyle="--", linewidth=2,
           label=f"99th pct (raw {q99_raw:.0f})")

# c) Labels & title
ax.set_xlabel("ln(connector_length_variance)")
ax.set_ylabel("Proportion of Trips")
ax.set_title(
    "Histogram of ln(connector-length variance)\n"
    "(non-zero trips; verticals = 90/95/99% raw cutoffs)"
)
ax.legend(title="Percentiles")

plt.tight_layout()
plt.show()
```
